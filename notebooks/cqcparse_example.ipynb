## CQCParse example

```python
from CQCParse.parsing import gaussian_parser as g16p
from CQCParse.parsing import cfour_parser as c4p
import logging
from CQCParse.logger import setup_logger
setup_logger("CQCParse", level=logging.INFO)
```

```python
logfile = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/dftGaussian/FORM/B3LYPcc_pVQZ/g16_inputFull_3q.out'
FORM_g16_1 = {'molecule': 'FORM', 'method': 'B3LYP', 'basis': 'cc-pVQZ', 
              'program': 'gaussian', 'log_file': logfile}
form_out1 = g16p.GaussianOutput(**FORM_g16_1)
form_out1_parser = g16p.GaussianParser(form_out1)
[i for i in form_out1_parser.__dir__() if '__' not in i]
form_out1_parser.load()
parsed1 = form_out1_parser.parse()

parsed1.vib_states.harmonic_states
parsed1.vib_states.anharmonic_states

parsed1.derivatives.__dict__.keys() # 'dipgrad', 'diphess', 'polgrad', 'polhess', 'cff', 'qff', 'cubic_cm_1', 'quartic_cm_1'
# parsed1.derivatives.cff # au unit

parsed1.normal_modes.normal_modes # array , units?

parsed1.anharm_correction_data.__dict__.keys() # 'rotational_constants', 'coriolis_constants', 'fermi_resonance', 'DD11', 'DD13', 'DD22'
# parsed1.anharm_treatment
# parsed1.list2exclude
```

```python
out_file = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/refinedc4/CCSDTcc_pVQZ/out'
polar_file = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/refinedc4/CCSDTcc_pVQZ/polar.pkl'
dipole_file = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/refinedc4/CCSDTcc_pVQZ/dipole'
cubic_file = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/refinedc4/CCSDTcc_pVQZ/cubic'
quartic_file = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/refinedc4/CCSDTcc_pVQZ/quartic'
outout_file = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/refinedc4/CCSDTcc_pVQZ/outfile0.out'
molden_f = '/home/vlev/wilson-suite/wilson_intensities/tests/test_database/refinedc4/CCSDTcc_pVQZ/MOLDEN'

FORM_c4_1 = {'molecule': 'FORM', 'method': 'CCSD(T)', 'basis': 'cc-pVQZ', 
              'program': 'cfour', 
              'out_file': out_file, 
              'molden_file':molden_f,
              'cubic_file': cubic_file,
              'quartic_file': quartic_file,
              'dipole_file': dipole_file,
              'polar_pkl': polar_file,
              }

form_out2 = c4p.CFOUROutput(**FORM_c4_1)
form_out2_parser = c4p.CFOURParser(form_out2)
form_out2_parser.load()

parsed2 = form_out2_parser.parse()

parsed2.__dict__.keys()

parsed2.vib_states.harmonic_states
parsed2.vib_states.anharmonic_states

parsed2.derivatives.__dict__.keys() # 'dipgrad', 'diphess', 'polgrad', 'polhess', 'cff', 'qff', 'cubic_cm_1', 'quartic_cm_1'
parsed2.derivatives.cff # au unit

parsed2.normal_modes.normal_modes # array , units?

parsed2.anharm_correction_data.__dict__.keys() # 'rotational_constants', 'coriolis_constants', 'fermi_resonance', 'DD11', 'DD13', 'DD22'
# parsed2.anharm_correction_data.rotational_constants
# parsed2.anharm_correction_data.coriolis_constants
# parsed2.anharm_treatment
# parsed2.list2exclude
parsed2.__dict__.keys() # 'rotational_constants', 'coriolis_constants', 'fermi_resonance', 'DD11', 'DD13', 'DD22'
```
