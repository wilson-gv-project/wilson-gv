```python
import functools

def function_call_counter(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.count += 1
        # print(f"-----> Function '{func.__name__}' called {wrapper.count} times.")
        print(f"-----> #{wrapper.count} - function '{func.__name__}' call")
        print(f"-----> Arguments: {args}, {kwargs}")
        return func(*args, **kwargs)
    
    wrapper.count = 0
    return wrapper

def my_function(arg1, arg2):
    # Function implementation goes here
    print('hii '+arg1)
    pass

def another_function():
    # Function implementation goes here
    print('hello')
    pass


# Wrap the functions with the function_call_counter wrapper
my_function = function_call_counter(my_function)
another_function = function_call_counter(another_function)

# Call the functions as usual
my_function("Hello", arg2=42)
my_function("World", arg2=24)
another_function()
```

```python
def factorial(x):
    """This is a recursive function
    to find the factorial of an integer"""

    if x == 1:
        return 1
    else:
        return (x * factorial(x-1))


# num = 3
# print("The factorial of", num, "is", factorial(num))
factorial = function_call_counter(factorial)
factorial(5)

```

```python
import numpy as np

# Assuming you have a function `my_function` that acts on meshgrid X and Y
def my_function(X, Y):
    # ... function implementation ...
    return X+Y

# Define the desired values of x and y
desired_x = 2.5
desired_y = 1.0

# Create the meshgrid
x = np.linspace(0, 5, 11)
y = np.linspace(0, 2, 5)
X, Y = np.meshgrid(x, y)

# Call your function with the meshgrid
result = my_function(X, Y)

# Find the indices that correspond to the desired values of x and y
x_index = np.where(x == desired_x)[0][0]
y_index = np.where(y == desired_y)[0][0]

# Access the corresponding element in the result matrix
desired_element = result[y_index, x_index]

# Print the desired element
print(desired_element)
print(result)
print(x, y)
```
